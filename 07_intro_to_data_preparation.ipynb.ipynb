# Data Preparation and Data Cleaning

You may have heard that **Data Scientists and Machine Learning Engineers spend a large amount of their time cleaning and preparing data**. This is true because real-world data is often messy, incomplete, inconsistent, or in the wrong format.

A common principle in Machine Learning is:

> **“Good models come from good data.”**

Even a powerful Machine Learning algorithm may perform poorly if the input data is not properly prepared.

## What is Data Preparation?

**Data preparation** is the process of cleaning, transforming, and preparing raw data before using it for analysis or Machine Learning.

Depending on the dataset and the problem, we may need to **remove data, modify existing features, or create new features**.

### Common Data Preparation Tasks

### 1. Remove Duplicate Values

Sometimes the same record appears multiple times in a dataset.

**Example:**

| Customer_ID | Name  | Age |
| ----------- | ----- | --: |
| 101         | Rahul |  25 |
| 102         | Amit  |  30 |
| 101         | Rahul |  25 |

The first and third rows are duplicates.

We can remove them using:

```python
df = df.drop_duplicates()
```

---

### 2. Handle Missing Values

Real-world datasets often contain missing values.

**Example:**

| Name  | Age | Salary |
| ----- | --: | -----: |
| Rahul |  25 |  40000 |
| Amit  | NaN |  50000 |
| Priya |  28 |    NaN |

We need to decide what to do with these missing values.

Possible approaches:

* Remove the rows.
* Fill missing values with mean, median, or mode.
* Use another suitable value.
* Leave them as they are if the algorithm can handle missing values.

For example:

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

The correct approach depends on the **dataset size, percentage of missing values, feature importance, and business problem**.

---

### 3. Remove Unnecessary or Noisy Features

Some features may not provide useful information for predicting the target.

**Example:**

Suppose we want to predict whether a customer will take a loan:

```text
Age
Income
Credit_Score
Loan_Amount
Customer_Name
Customer_ID
```

`Customer_Name` may not help the model predict loan approval.

Therefore, we may remove unnecessary features:

```python
df = df.drop(columns=["Customer_Name"])
```

However, we should not remove a feature only because it looks unnecessary. We should analyze its usefulness first.

### 5. Encode Categorical Features

Machine Learning models generally require numerical input, so categorical values often need to be converted into numbers.

**Example:**

```text
Gender
------
Male
Female
Male
Female
```

We can encode them as:

```text
Male   → 1
Female → 0
```

For multiple categories, we can use techniques such as **One-Hot Encoding**.

Example:

```text
City
----
Pune
Mumbai
Delhi
```

can become:

```text
City_Pune  City_Mumbai  City_Delhi
1          0            0
0          1            0
0          0            1
```

### 6. Create New Features

Sometimes existing features do not provide enough useful information. We can create new features from existing data. This is called **Feature Engineering**.

**Example:**

Suppose we have:

```text
Date_of_Birth
```

Instead of directly using the date, we can calculate:

```text
Age
```

Another example:

```text
Total_Amount = Quantity × Price
```

```python
df["Total_Amount"] = df["Quantity"] * df["Price"]
```

Good feature engineering can improve the performance of a Machine Learning model, especially when the new features have strong predictive power.

---

### 7. Standardize or Normalize Numerical Features

Numerical features can have very different scales.

**Example:**

```text
Age       → 18 to 80
Salary    → 20,000 to 2,00,000
```

Some Machine Learning algorithms can be affected by these differences in scale.

#### Normalization

Normalization usually scales values to a range between **0 and 1**.

Example:

```text
10 → 0.0
50 → 0.5
100 → 1.0
```

#### Standardization

Standardization transforms the data so that it has approximately:

```text
Mean = 0
Standard Deviation = 1
```

For example, StandardScaler can be used:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df[["Age", "Salary"]])
```

Scaling is particularly important for algorithms such as **KNN, K-Means, SVM, and many gradient-based models**.

---

## 8. Handle Outliers

Outliers are values that are unusually high or low compared with the rest of the data.

**Example:**

```text
Salary:
30,000
35,000
40,000
42,000
5,00,00,000
```

`5,00,00,000` may be an outlier.

We need to investigate whether it is:

* A genuine value
* A data-entry error
* A special case

Depending on the situation, we may keep, transform, cap, or remove the outlier.

---

## 9. Split the Dataset

After preparing the data, we usually divide it into different datasets.

For example:

```text
Training Data → Used to train the model
Validation Data → Used to tune/select the model
Test Data → Used to evaluate final performance
```

A common split could be:

```text
Training → 70%
Validation → 15%
Test → 15%
```

The exact split depends on the project.

---

## 10. Shuffle the Data

Sometimes the data is arranged in a particular order.

**Example:**

```text
Rows 1–500   → Class A
Rows 501–1000 → Class B
```

If we directly split this data, the training and test sets may not represent the overall dataset properly.

Shuffling helps distribute the observations more randomly.

```python
df = df.sample(frac=1, random_state=42)
```

# Important Point

The above tasks are **not a fixed step-by-step checklist** that must always be followed in exactly the same order.

The required data preparation steps depend on:

* Dataset
* Business problem
* Machine Learning algorithm
* Data quality
* Feature types
* Missing values
* Target variable

For example, one dataset may require missing-value treatment, while another dataset may have no missing values.

## Simple Example

Suppose we have a customer dataset:

| Age | Salary | Gender | City   | Loan_Status |
| --: | -----: | ------ | ------ | ----------- |
|  25 |  40000 | Male   | Pune   | Yes         |
| NaN |  50000 | Female | Mumbai | No          |
|  30 |  45000 | Male   | Pune   | Yes         |
|  30 |  45000 | Male   | Pune   | Yes         |

Before building a model, we might:

1. Remove the duplicate row.
2. Handle the missing `Age`.
3. Encode `Gender`.
4. Encode `City`.
5. Check for outliers in `Salary`.
6. Select useful features.
7. Scale numerical features if required.
8. Split the data into training, validation, and test sets.

### In Simple Words

**Raw Data → Clean Data → Transform Data → Create Useful Features → Prepare Data → Train ML Model**

The quality of data preparation has a major impact on the quality and reliability of a Machine Learning model.
