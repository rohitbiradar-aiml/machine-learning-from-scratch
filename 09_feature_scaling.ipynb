# A Handy Note About Feature Scaling

**Feature Scaling** is a preprocessing technique used to bring numerical features into a **similar range or scale**.

In real-world datasets, different features may have very different ranges. For example, `age` may range from 1–80, while `income` may range from 10,000–1,00,000. Such differences in scale can affect the performance of some machine learning algorithms.

Feature scaling helps ensure that features with larger numerical values do not **dominate features with smaller values**.

Some commonly used feature scaling techniques are:

* **Standardization**
* **Min-Max Scaling (Normalization)**
* **Robust Scaling**

Let's understand and implement these techniques.


# A Handy Note About Feature Scaling

Machine Learning models can be selective about the **type and range of values** provided as input. Except for some models, such as **Decision Trees**, many machine learning algorithms perform better when the input features are properly scaled.

### What is Feature Scaling?

**Feature Scaling** is the process of transforming numerical features into a **similar or smaller range of values**.

For example, suppose a dataset contains:

* `Age` → values between 1 and 80
* `Salary` → values between 10,000 and 1,00,000

Because these features have very different ranges, some machine learning algorithms may give more importance to the feature with larger values. Feature scaling helps bring them to a comparable scale.

In this notebook, we will explore the following feature scaling techniques:

* **Normalization**
* **Standardization**
* **Robust Scaling**

### Normalization vs Standardization

**Normalization** and **Standardization** are different feature scaling techniques. Although they can sometimes be used interchangeably, they work differently and are suitable for different situations.

We will understand the difference between these techniques and implement each of them.


## 1. Normalization

Normalization is a scaling techniques that transform the numerical feature to the range of values between 0 and 1.

Here is a formula that is followed when normalizing the data. $Xmin$ is the minimum value of feature X, and $Xmax$ is the maximum value of X.



$$
Xnorm = \frac {X-Xmin} {Xmax-Xmin}
$$

#### When Should you Normalize the Features?

#### When Should You Normalize the Features?

When features have **different ranges of values**, normalization can be a useful scaling technique.

For example, suppose we have two features:

* Feature 1 → values from **1 to 100**
* Feature 2 → values from **5 to 300**

Since these features have different ranges, we may need to scale them to bring them into a **similar range**.

### When is Normalization Preferred?

Normalization is generally preferred when the data **does not follow a normal (Gaussian) distribution**.

If the data follows a **normal or Gaussian distribution**, **Standardization** is often a better choice.

If we do not know the distribution of the data, normalization can be considered as an initial approach, depending on the machine learning algorithm and dataset.

### Normalization with Different Algorithms

Normalization can be particularly useful for algorithms that are sensitive to the scale of input features, such as:

* **Neural Networks**
* **K-Nearest Neighbors (KNN)**

For example, KNN calculates distances between data points. Therefore, features with larger ranges can have a greater influence on the distance calculation. Scaling the features helps prevent this problem.

In general, the choice of scaling technique should depend on the **data distribution, algorithm, and characteristics of the dataset**.


Most popular ML frameworks provide functions to normalize the numerical data.

For illustration purpose, I will use tips data available in Seaborn.

````python
from seaborn import load_dataset

tip_data = load_dataset('tips')
tip_data.head()
````

Let's take all numerical features from the above data.

````python
num_feats = tip_data[['total_bill', 'tip', 'size']]
````

For now let's scale those numerical features with Scikit-Learn preprocessing functions. We will use `MinMaxScaler` which scale the data to the range between 0 and 1 by default. If you want a different range, you can change that.

````python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

num_scaled = scaler.fit_transform(num_feats)
num_scaled[:5]
````

The output of the scaler is a NumPy array. You can convert it back to a Pandas DataFrame.

````python
import pandas as pd

num_scaled_df = pd.DataFrame(num_scaled, columns=num_feats.columns)
num_scaled_df.head()
````

As you can see above, all the values are scaled to the values between 0 and 1.

## 2. Standardization

In standardization, the numerical features are rescaled to have the 0 mean($u$) and unity standard deviation(std or $\sigma$ ).

Here is the formula of standardization. Xstd is the standardized feature, X is the feature, $u$ is mean of the feature, and $\sigma$ is the standard deviation.

$$
Xstd = \frac {X - u} {\sigma}
$$

#### When Should you Standardize the Features?

When you know that the training data at hand has a normal or gaussian distribution, you should standardize such data.

Some ML algorithms such as Support Vector Machines(with rbf kernel) and linear models expect that the input data to have a normal distribution.

In most cases, whether you choose normalization or standardization, it won't make much difference, but it can. So, it makes sense to try both especially if you are not sure about the distribution of the data.

Here is how Standardization is implemented in Scikit-Learn.

````python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()
num_std = std_scaler.fit_transform(num_feats)
num_std[:5]
````

````python
# The mean of each feature in the scaled data

std_scaler.mean_
````

````python
# Variance of the scaled features

std_scaler.var_
````

Scaled data has zero mean an unit variance.

````python
import numpy as np

print(f'The mean of scaled data: {np.round(num_std.mean(axis=0))}')
print(f'The standard deviation of scaled data: {num_std.std(axis=0)}')
````

````python
# Converting the scaled values back to datframe

num_std_scaled_df = pd.DataFrame(num_std, columns=num_feats.columns)
num_std_scaled_df.head()
````

## 3. Robust Scaler

[Robust scaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.RobustScaler.html#sklearn.preprocessing.RobustScaler) is kind of similar to standardization but is used when the data contains many outliers.

Instead of dropping mean, the median is dropped and the data is scaled to the Interquartile Range(IQR). The IQR is the range between the 1st quartile (25th quantile) and the 3rd quartile (75th quantile).

Like normalization and standardization, Roust scaler is also implemented easily in Scikit-Learn.

````python
from sklearn.preprocessing import RobustScaler

rob_scaler = RobustScaler()
num_rob_scaled = rob_scaler.fit_transform(num_feats)

num_rob_scaled[:5]
````

By scaling the data with Robust Scaler, the median of the resulting values will have a median of zero.

````python
print(f'The median of scaled data: {np.round(np.median(num_rob_scaled, axis=0))}')
````

### Final Notes

### Important Points About Feature Scaling

Scaling the input features before feeding them into a machine learning model is often a good practice, especially for algorithms that are sensitive to the scale of the input data.

Here are the key points to remember:

* **Scaling** the features can help some machine learning models **converge faster** and perform better.

* **Normalization** scales the data to a fixed range, commonly **0 to 1**. It is often preferred when the data does not follow a normal distribution.

* **Standardization** transforms the data so that it has a **mean of 0 and a standard deviation of 1**. It is commonly preferred when the data approximately follows a normal or Gaussian distribution.

* **Robust Scaling** is useful when the dataset contains **many outliers**, because it is less affected by extreme values.

* There is no single scaling technique that is always best. The choice depends on the **dataset and machine learning algorithm**. In practice, it can be useful to try different techniques and evaluate which one works best.

* Usually, we scale the **input features (`X`)**, not the target labels (`y`) for a classification problem.

### Important: Avoid Data Leakage

The scaler should be **fitted only on the training data**. We should never fit the scaler on the test data because this can cause **data leakage**.

**Don't:**

```python
scaler.fit_transform(X_test)
```

**Do:**

```python
scaler.transform(X_test)
```

The correct workflow is:

```python
scaler.fit_transform(X_train)
scaler.transform(X_test)
```

Here, the scaler learns the scaling parameters from `X_train` and then uses the **same parameters** to transform `X_test`.


**In short, choose the scaling technique based on your data and machine learning model, and always fit the scaler only on the training data before transforming the test data.**
